#### re

```python
import re
```

```python
str1='Welcome to python programming ^&&%^%^%$)()**?<> July 2026 class'
```

```python
re.findall('python',str1)
```

```python
re.search('python',str1)
```

```python
#### \d--extract only numbers
re.findall('\d+',str1)
```

```python
#### \D--extract except numbers
re.findall('\D+',str1)
```

```python
#### \s--extract only spaces
re.findall('\s+',str1)
```

```python
#### \S--extract except spaces
re.findall('\S+',str1)
```

```python
#### \w--extract except spaces and Non-Alphanumeric
re.findall('\w+',str1)
```

```python
#### \W--extract only spaces and Non-Alphanumeric
re.findall('\W+',str1)
```
